```python
# Create random array

arr = np.random.randint(1, 101, 20)

print("Original Array:")
print(arr)
```

```python
mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)
minimum = np.min(arr)
maximum = np.max(arr)

print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
matrix = arr.reshape(4,5)

print("4 x 5 Matrix")
print(matrix)
```

```python
print("First Two Rows")
print(matrix[:2])
```

```python
multiplied_matrix = matrix * 2

print("Matrix After Multiplying by 2")
print(multiplied_matrix)
```

```python
with open("../outputs/numpy_results.txt","w") as file:

    file.write("Original Array\n")
    file.write(str(arr))

    file.write("\n\nMean: " + str(mean))
    file.write("\nMedian: " + str(median))
    file.write("\nStd: " + str(std))
    file.write("\nMin: " + str(minimum))
    file.write("\nMax: " + str(maximum))

print("NumPy Results Saved")
```

# Dataset Loading

Titanic Dataset is loaded using Pandas.

```python
df = pd.read_csv("../data/titanic.csv")

df.head()
```

```python
df.head(10)
```

```python
print("Shape of Dataset")
print(df.shape)
```

```python
print(df.dtypes)
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
missing = pd.DataFrame(
    df.isnull().sum(),
    columns=["Missing Values"]
)

missing
```

```python
df["Age"] = df["Age"].fillna(
    df["Age"].mean()
)
```

```python
df["Embarked"] = df["Embarked"].fillna(
    "Unknown"
)
```

```python
df["Cabin"] = df["Cabin"].fillna(
    "Not Available"
)
```

```python
before = df.shape[0]

df = df.drop_duplicates()

after = df.shape[0]

print("Rows Before:", before)
print("Rows After:", after)
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].astype(float)

print(df["Age"].dtype)
```

```python
df.to_csv(
    "../data/cleaned_titanic.csv",
    index=False
)

print("Clean Dataset Saved")
```

# Data Merging

Creating a supplementary dataframe and merging it with Titanic dataset.

```python
class_df = pd.DataFrame({

    "Pclass":[1,2,3],

    "Class_Name":[
        "First Class",
        "Second Class",
        "Third Class"
    ]
})

class_df
```

```python
merged_df = pd.merge(

    df,
    class_df,

    on="Pclass",
    how="left"
)
```

```python
merged_df.head()
```

```python
merged_df.to_csv(
    "../outputs/merged_dataset.csv",
    index=False
)

print("Merged Dataset Saved")
```

```python
with open(
    "../outputs/cleaning_report.txt",
    "w"
) as file:

    file.write("DATA CLEANING REPORT\n")
    file.write("===================\n\n")

    file.write("1. Missing values handled.\n")
    file.write("2. Age column filled using mean.\n")
    file.write("3. Embarked column filled with Unknown.\n")
    file.write("4. Cabin column filled with Not Available.\n")
    file.write("5. Duplicate rows removed.\n")
    file.write("6. Data types verified.\n")

print("Cleaning Report Saved")
```
