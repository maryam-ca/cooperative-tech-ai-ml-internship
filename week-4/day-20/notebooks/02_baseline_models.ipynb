```python
# ============================
# Import Required Libraries
# ============================

import os
import joblib
import warnings

import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")
```

# Load Processed Dataset

The processed dataset created in Notebook 1 is loaded for model training and evaluation.

```python
# ============================
# Load Processed Data
# ============================

X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

y_train = pd.read_csv("../data/processed/y_train.csv")["target"]
y_test = pd.read_csv("../data/processed/y_test.csv")["target"]

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)
```

# Create Results Storage

A list is created to store the evaluation metrics of all baseline models.

```python
# ============================
# Results Storage
# ============================

results = []
```

# Logistic Regression

Logistic Regression is one of the most commonly used baseline classification algorithms. It is simple, interpretable, and computationally efficient.

```python
# ============================
# Logistic Regression
# ============================

logreg = LogisticRegression(random_state=42)

logreg.fit(X_train, y_train)

logreg_pred = logreg.predict(X_test)
```

```python
# Logistic Regression Metrics

logreg_accuracy = accuracy_score(y_test, logreg_pred)
logreg_precision = precision_score(y_test, logreg_pred)
logreg_recall = recall_score(y_test, logreg_pred)
logreg_f1 = f1_score(y_test, logreg_pred)

print(classification_report(y_test, logreg_pred))
```

```python
results.append({
    "Model": "Logistic Regression",
    "Accuracy": logreg_accuracy,
    "Precision": logreg_precision,
    "Recall": logreg_recall,
    "F1 Score": logreg_f1
})
```

# Decision Tree Classifier

Decision Trees learn decision rules from the training data and are widely used as baseline models because they are easy to understand and visualize.

```python
# ============================
# Decision Tree
# ============================

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train, y_train)

dt_pred = dt.predict(X_test)
```

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
dt_precision = precision_score(y_test, dt_pred)
dt_recall = recall_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred)

print(classification_report(y_test, dt_pred))
```

```python
results.append({
    "Model": "Decision Tree",
    "Accuracy": dt_accuracy,
    "Precision": dt_precision,
    "Recall": dt_recall,
    "F1 Score": dt_f1
})
```

# Random Forest Classifier

Random Forest combines multiple decision trees to improve prediction accuracy and reduce overfitting. It serves as a stronger baseline model.

```python
# ============================
# Random Forest
# ============================

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
```

```python
rf_accuracy = accuracy_score(y_test, rf_pred)
rf_precision = precision_score(y_test, rf_pred)
rf_recall = recall_score(y_test, rf_pred)
rf_f1 = f1_score(y_test, rf_pred)

print(classification_report(y_test, rf_pred))
```

```python
results.append({
    "Model": "Random Forest",
    "Accuracy": rf_accuracy,
    "Precision": rf_precision,
    "Recall": rf_recall,
    "F1 Score": rf_f1
})
```

# Model Comparison

The evaluation metrics of all baseline models are combined into a single comparison table.

```python
baseline_results = pd.DataFrame(results)

baseline_results
```

```python
baseline_results.sort_values(
    by="Accuracy",
    ascending=False
)
```

# Save Baseline Models

The trained models are saved for future prediction and comparison with advanced models.

```python
joblib.dump(logreg, "../models/baseline_logreg.pkl")

joblib.dump(dt, "../models/baseline_dt.pkl")

joblib.dump(rf, "../models/baseline_rf.pkl")

print("Baseline models saved successfully.")
```

# Save Evaluation Results

The evaluation metrics are exported as a CSV file for use in the final model comparison notebook.

```python
os.makedirs("../outputs/model_results", exist_ok=True)

baseline_results.to_csv(
    "../outputs/model_results/baseline_metrics.csv",
    index=False
)

print("Baseline metrics saved successfully.")
```
