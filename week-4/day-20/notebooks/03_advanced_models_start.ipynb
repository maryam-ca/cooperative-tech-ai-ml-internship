### Objective

This notebook focuses on training advanced machine learning models using the preprocessed Bank Marketing dataset.

The following tasks will be performed:

- Load processed data
- Train Gradient Boosting
- Train XGBoost
- Train Support Vector Machine (SVM)
- Evaluate each model
- Compare model performance
- Save trained models
- Export evaluation metrics

These models will later be compared with the baseline models to identify the best-performing algorithm.

```python
# ============================
# Import Required Libraries
# ============================

import os
import joblib
import warnings

import pandas as pd

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

import xgboost as xgb

warnings.filterwarnings("ignore")
```

# Load Processed Dataset

The processed training and testing datasets created in Notebook 1 are loaded for advanced model training.

```python
# ============================
# Load Dataset
# ============================

X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

y_train = pd.read_csv("../data/processed/y_train.csv")["target"]
y_test = pd.read_csv("../data/processed/y_test.csv")["target"]

print(X_train.shape)
print(X_test.shape)
```

# Create Results Storage

A list is created to store the evaluation metrics of all advanced models.

```python
results = []
```

# Gradient Boosting Classifier

Gradient Boosting builds trees sequentially, where each tree attempts to correct the mistakes of the previous one.

```python
gb = GradientBoostingClassifier(random_state=42)

gb.fit(X_train, y_train)

gb_pred = gb.predict(X_test)
```

```python
gb_accuracy = accuracy_score(y_test, gb_pred)
gb_precision = precision_score(y_test, gb_pred)
gb_recall = recall_score(y_test, gb_pred)
gb_f1 = f1_score(y_test, gb_pred)

print(classification_report(y_test, gb_pred))
```

```python
results.append({
    "Model":"Gradient Boosting",
    "Accuracy":gb_accuracy,
    "Precision":gb_precision,
    "Recall":gb_recall,
    "F1 Score":gb_f1
})
```

# XGBoost Classifier

XGBoost is an optimized implementation of Gradient Boosting that is widely used in industry because of its speed and predictive performance.

```python
xgb_model = xgb.XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)
```

```python
xgb_accuracy = accuracy_score(y_test, xgb_pred)
xgb_precision = precision_score(y_test, xgb_pred)
xgb_recall = recall_score(y_test, xgb_pred)
xgb_f1 = f1_score(y_test, xgb_pred)

print(classification_report(y_test, xgb_pred))
```

```python
results.append({
    "Model":"XGBoost",
    "Accuracy":xgb_accuracy,
    "Precision":xgb_precision,
    "Recall":xgb_recall,
    "F1 Score":xgb_f1
})
```

# Support Vector Machine (SVM)

Support Vector Machine is a supervised learning algorithm that identifies the optimal decision boundary for separating different classes.

```python
svm = SVC(random_state=42)

svm.fit(X_train, y_train)

svm_pred = svm.predict(X_test)
```

```python
svm_accuracy = accuracy_score(y_test, svm_pred)
svm_precision = precision_score(y_test, svm_pred)
svm_recall = recall_score(y_test, svm_pred)
svm_f1 = f1_score(y_test, svm_pred)

print(classification_report(y_test, svm_pred))
```

```python
results.append({
    "Model":"Support Vector Machine",
    "Accuracy":svm_accuracy,
    "Precision":svm_precision,
    "Recall":svm_recall,
    "F1 Score":svm_f1
})
```

# Advanced Model Comparison

The performance of all advanced machine learning models is summarized in a comparison table.

```python
advanced_results = pd.DataFrame(results)

advanced_results
```

```python
advanced_results.sort_values(
    by="Accuracy",
    ascending=False
)
```

# Save Trained Models

The trained advanced models are saved for future use and deployment.

```python
joblib.dump(gb,"../models/gradient_boosting.pkl")

joblib.dump(xgb_model,"../models/xgboost.pkl")

joblib.dump(svm,"../models/svm.pkl")

print("Advanced models saved successfully.")
```

# Save Evaluation Results

The evaluation metrics of all advanced models are exported as a CSV file for comparison in the next notebook.

```python
os.makedirs("../outputs/model_results",exist_ok=True)

advanced_results.to_csv(
    "../outputs/model_results/advanced_metrics.csv",
    index=False
)

print("Advanced metrics saved successfully.")
```
