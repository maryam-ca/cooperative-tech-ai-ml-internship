### Objective

The objective of this notebook is to compare the performance of all baseline and advanced machine learning models.

The following tasks will be performed:

- Load evaluation metrics
- Merge baseline and advanced model results
- Compare model performance
- Visualize model metrics
- Identify the best-performing model
- Save comparison results

This notebook completes the Day 20 workflow.

```python
# ============================
# Import Required Libraries
# ============================

import os
import warnings

import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")
```

# Load Model Evaluation Results

The evaluation metrics generated in the previous notebooks are loaded for comparison.

```python
# ============================
# Load Baseline Metrics
# ============================

baseline = pd.read_csv("../outputs/model_results/baseline_metrics.csv")

baseline
```

```python
# ============================
# Load Advanced Metrics
# ============================

advanced = pd.read_csv("../outputs/model_results/advanced_metrics.csv")

advanced
```

# Merge Evaluation Results

The baseline and advanced model metrics are combined into a single comparison table.

```python
comparison = pd.concat(
    [baseline, advanced],
    ignore_index=True
)

comparison
```

# Sort Models by Accuracy

The models are ranked according to their classification accuracy.

```python
comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
)

comparison
```

# Best Performing Model

The model with the highest accuracy is identified.

```python
best_model = comparison.iloc[0]

print("Best Model")
print("---------------------")
print(best_model)
```

# Accuracy Comparison

A bar chart is created to compare the accuracy of all machine learning models.

```python
plt.figure(figsize=(10,6))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title("Model Accuracy Comparison")
plt.xlabel("Models")
plt.ylabel("Accuracy")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()
```

# F1-Score Comparison

The F1-score provides a balanced evaluation by considering both precision and recall.

```python
plt.figure(figsize=(10,6))

plt.bar(
    comparison["Model"],
    comparison["F1 Score"]
)

plt.title("Model F1-Score Comparison")
plt.xlabel("Models")
plt.ylabel("F1 Score")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()
```

# Precision and Recall Comparison

Precision and Recall are compared to evaluate the classification quality of each model.

```python
comparison.plot(
    x="Model",
    y=["Precision","Recall"],
    kind="bar",
    figsize=(10,6)
)

plt.title("Precision vs Recall")

plt.ylabel("Score")

plt.tight_layout()

plt.show()
```

# Save Comparison Results

The combined evaluation metrics are exported for future reference.

```python
os.makedirs("../outputs/model_results", exist_ok=True)

comparison.to_csv(
    "../outputs/model_results/model_comparison.csv",
    index=False
)

print("Comparison table saved successfully.")
```
